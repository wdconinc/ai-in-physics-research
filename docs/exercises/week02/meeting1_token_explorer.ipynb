# Week 2 Meeting 1 — Token Explorer

Starter notebook for tokenization, token counting, and basic API debugging.

This notebook uses the [Anthropic Python SDK](https://github.com/anthropics/anthropic-sdk-python).
Set the `ANTHROPIC_API_KEY` environment variable before running the API cells.


```python
import os
import anthropic

sentence = "Iron-56 has one of the highest binding energies per nucleon in nature."

# Anthropic does not expose individual token IDs publicly, but the SDK
# provides a token-counting endpoint so you can see how the model will
# charge for a given piece of text.
client = anthropic.Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))
response = client.messages.count_tokens(
    model="claude-3-5-haiku-20241022",
    messages=[{"role": "user", "content": sentence}],
)

print(f"Sentence : {sentence!r}")
print(f"Token count (claude-3-5-haiku-20241022): {response.input_tokens}")
```

```python
import os
import anthropic

def count_tokens(prompt: str, model: str = "claude-3-5-haiku-20241022") -> int:
    """Return the number of input tokens Anthropic would charge for *prompt*."""
    client = anthropic.Anthropic(api_key=os.environ.get("ANTHROPIC_API_KEY"))
    response = client.messages.count_tokens(
        model=model,
        messages=[{"role": "user", "content": prompt}],
    )
    return response.input_tokens

print(count_tokens(""))
print(count_tokens("Compute the de Broglie wavelength of a 200 keV electron."))
```
