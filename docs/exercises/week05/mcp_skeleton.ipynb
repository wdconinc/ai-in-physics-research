```python
# --- Quick unit tests (run before connecting the LLM client) ---
import asyncio

async def _test_simbad():
    result = await query_simbad("Betelgeuse")
    assert "spectral_type" in result, "Missing spectral_type key"
    print("SIMBAD result:", json.dumps(result, indent=2))

async def _test_arxiv():
    results = await search_arxiv("Betelgeuse", max_results=3)
    assert isinstance(results, list) and len(results) > 0, "Expected a non-empty list"
    print(f"arXiv: {len(results)} result(s). First title: {results[0]['title']}")

await _test_simbad()
await _test_arxiv()
```
