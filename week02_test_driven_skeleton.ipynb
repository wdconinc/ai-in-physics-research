```python
%%writefile test_projectile.py
import math
from projectile_solver import simulate_projectile

def test_vacuum_range():
    v0, theta = 30.0, 45.0
    r, _, _ = simulate_projectile(v0, theta, b=0.0)
    expected = v0**2 * math.sin(math.radians(2 * theta)) / 9.81
    assert math.isfinite(r)
    assert abs(r - expected) / max(expected, 1e-9) < 0.05

def test_vacuum_max_height():
    v0, theta = 30.0, 45.0
    _, h, _ = simulate_projectile(v0, theta, b=0.0)
    expected = (v0**2 * math.sin(math.radians(theta))**2) / (2 * 9.81)
    assert math.isfinite(h)
    assert abs(h - expected) / max(expected, 1e-9) < 0.05

def test_drag_reduces_range():
    v0, theta = 30.0, 45.0
    r_vac, _, _ = simulate_projectile(v0, theta, b=0.0)
    r_drag, _, _ = simulate_projectile(v0, theta, b=0.1)
    assert r_drag < r_vac

def test_time_of_flight_positive():
    _, _, t = simulate_projectile(25.0, 30.0, b=0.1)
    assert math.isfinite(t)
    assert t > 0

def test_symmetry_broken_by_drag():
    v0 = 30.0
    r35, _, _ = simulate_projectile(v0, 35.0, b=0.2)
    r45, _, _ = simulate_projectile(v0, 45.0, b=0.2)
    assert r35 > r45
```
