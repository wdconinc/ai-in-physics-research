Starter notebook for tokenization, token counting, and basic API debugging.


```python
import tiktoken

sentence = "Iron-56 has one of the highest binding energies per nucleon in nature."
enc = tiktoken.get_encoding("cl100k_base")
token_ids = enc.encode(sentence)
decoded = [enc.decode([tid]) for tid in token_ids]

for tid, tok in zip(token_ids, decoded):
    print(f"{tid:>6}  {tok!r}")
print(f"Total tokens: {len(token_ids)}")
```

```python
import tiktoken

def count_tokens(prompt: str, model: str = "gpt-4o-mini") -> int:
    """TODO: complete this function using tiktoken.encoding_for_model."""
    # Starter implementation for students to refine.
    try:
        enc = tiktoken.encoding_for_model(model)
    except KeyError:
        enc = tiktoken.get_encoding("cl100k_base")
    return len(enc.encode(prompt))

print(count_tokens(""))
print(count_tokens("Compute the de Broglie wavelength of a 200 keV electron."))
```
